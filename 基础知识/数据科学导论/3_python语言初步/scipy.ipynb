# Scipy
## 基本功能:保存和读取矩阵文件


```python
from scipy import io
import numpy as np

arr = np.array([[1, 2, 3], [4, 5, 6]])
io.savemat('array_data.mat', {'array': arr})

br = io.loadmat('array_data.mat')
print(br['array'])
```

## 统计功能
- 使用scipy中stats的函数可以创建服从某些分布的随机数

```python
from scipy import stats

# 正态分布
normal_dist = stats.norm(loc=0, scale=1)    # 均值为0，标准差为1的正态分布
random_samples = normal_dist.rvs(size=10)    # 生成10个随机样
print("正态分布随机样本：", random_samples)
# # 指数分布
# exponential_dist = stats.expon(scale=1)    # 均值为1的指数分布
# random_samples_exp = exponential_dist.rvs(size=10)    # 生成10个随机样本
# print("指数分布随机样本：", random_samples_exp)
# # 泊松分布
# poisson_dist = stats.poisson(mu=3)    # 均值为3的泊松分布
# random_samples_pois = poisson_dist.rvs(size=10)    # 生成10个随机样本
# print("泊松分布随机样本：", random_samples_pois)

# 利用stats.norm.fit方法对数据进行正态分布拟合
random_samples = normal_dist.rvs(size=1000)
mu, std = stats.norm.fit(random_samples)
print(f"拟合的正态分布参数：均值={mu}, 标准差={std}")

# 利用skewtest方法检验数据的偏度
skewness_test = stats.skewtest(random_samples)
normal_test = stats.normaltest(random_samples)
print(f"偏度检验结果：统计量={skewness_test.statistic}, p值={skewness_test.pvalue}")    #p值用于判断偏度是否显著,即数据服从正态分布的概率
print(f"正态性检验结果：统计量={normal_test.statistic}, p值={normal_test.pvalue}")      #p值用于判断数据是否服从正态分布
# 从结果来看,两个函数的p值不同,这是因为它们检验的内容不同,偏度检验只关注数据的偏度,而正态性检验则综合考虑了偏度和峰度两个方面
```
