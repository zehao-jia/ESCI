## Series
- series类似于np中的一维数组,只有一个栏,其下标可以随意更改




```python
import pandas as pd
from pandas import Series,DataFrame

obj_1 = Series([2,5,-8,1])
print(obj_1)
obj_2 = Series([2,5,6,7],index=['a','b','c','d'])
print(obj_2)

obj_2['a'] = 10
print(obj_2)
```

## DataFrame数据结构
- dataframe这一数据结构可以有不同的列,每列可以有多个数据类型


```python
data = {'name':['张三','李四','王五','赵六'],
        'age':[23,24,22,21],
        'gender':['男','男','男','女']}
frame = DataFrame(data)
print(frame)
print(frame['name'])
print(frame['age'][2])
```

- 指定列顺序,则根据列顺序排列

```python
frame2 = DataFrame(data,columns=['name','gender','age'])
print(frame2)
```

```python
## pandas的几个基本操作
```

```python
frame3 = DataFrame(data,columns=['name','gender','age'],index=['a','b','c','d'])
frame3 = frame3.reindex(['a','b','c','d','e'])        # 重新指定行索引
print(frame3)
frame3 = frame3.drop('a')                             # 删除指定行
print(frame3)
print(frame3.loc['b':'c','name':'gender'])            # 按标签选取数据
print(frame3.iloc[0:3,1:2])                           # 按位置选取数据
```
