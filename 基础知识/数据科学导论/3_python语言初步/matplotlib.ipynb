# Matplotlib
- matplotlib的图像都存在创建的figure对象之中

```python
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111)
ax.set(xlim=[0.5, 4.5], ylim=[0, 5], title='Sample Plot', xlabel='X-axis', ylabel='Y-axis')
x = np.linspace(1, 4, 100)
y = np.sin(8*x) + 2.5
ax.plot(x, y, label='Sine Wave', color='blue')
plt.legend('blue')

plt.show()
```
